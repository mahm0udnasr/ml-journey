# random module
- Generates **pseudo-random** numbers: random-looking, but produced by an algorithm from a starting value (the **seed**).
- Used for shuffling data, train/test splits, sampling, simulations, data augmentation...
- ⚠️ Not for passwords or security tokens — use the `secrets` module for that.

### Most used ⭐
| Function | Returns |
|---|---|
| `random.seed(n)` | fix the seed → same "random" results every run (**reproducibility** 🔑 in ML) |
| `random.random()` | float in `[0.0, 1.0)` |
| `random.uniform(a, b)` | float between a and b |
| `random.randint(a, b)` | int between a and b, **both included** |
| `random.randrange(start, stop, step)` | int like `range()` (stop excluded) |
| `random.choice(seq)` | one random item |
| `random.choices(seq, weights, k)` | k items **with** replacement (can repeat) |
| `random.sample(seq, k)` | k **unique** items (no replacement) |
| `random.shuffle(lst)` | shuffles the list **in place**, returns `None` |
| `random.gauss(mu, sigma)` | number from a normal (bell curve) distribution |

````python
import random
````

### 1. Seed & reproducibility ⭐
Same seed → same sequence. In ML you set seeds so experiments can be repeated exactly.

````python
random.seed(42)
print([random.randint(1, 10) for _ in range(5)])

random.seed(42)
print([random.randint(1, 10) for _ in range(5)])   # identical!

random.seed()   # back to unpredictable
````

### 2. Random numbers

````python
random.seed(0)
print(random.random())
print(random.uniform(1.5, 3.5))
print(random.randint(1, 6))           # dice: 1..6 (6 included)
print(random.randrange(0, 100, 10))   # 0, 10, ..., 90
print(random.gauss(0, 1))             # mean 0, std 1
````

### 3. Picking from sequences

````python
random.seed(1)
colors = ["red", "green", "blue", "yellow"]

print(random.choice(colors))
print(random.choices(colors, k=6))                       # repeats allowed
print(random.choices(["win", "lose"], weights=[1, 9], k=10))
print(random.sample(colors, 2))                          # no repeats
print(random.sample(range(1, 50), 6))                    # lottery numbers
````

### 4. Shuffling
⚠️ `shuffle` changes the list itself and returns `None` — don't write `x = random.shuffle(x)`.
To get a shuffled **copy**, use `random.sample(lst, len(lst))`.

````python
random.seed(3)
deck = list(range(1, 11))
result = random.shuffle(deck)
print(deck, result)

original = ["a", "b", "c", "d"]
shuffled = random.sample(original, len(original))
print(original, shuffled)
````

**Estimate π (Monte Carlo)** — throw 100,000 random points `(x, y)` with x, y in `[0, 1)`.
The fraction that falls inside the circle `x² + y² <= 1` is about π/4. Print your estimate of π.

````python
# your code here
````

<details>
<summary>💡 Show solution</summary>

```python
n = 100_000
inside = 0
for _ in range(n):
    x, y = random.random(), random.random()
    if x * x + y * y <= 1:
        inside += 1
print(4 * inside / n)
```
