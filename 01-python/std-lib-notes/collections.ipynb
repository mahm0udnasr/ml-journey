# collections module
- `collections` gives you **specialized containers** — upgraded versions of `dict`, `tuple` and `list` for common tasks.

### Most used ⭐
| Class | What it is | Use it for |
|---|---|---|
| `Counter` | dict that counts things | word counts, label/class distribution |
| `defaultdict` | dict with a default value for missing keys | grouping, no more `if key not in d` |
| `namedtuple` | tuple with named fields | small readable records (`p.x` instead of `p[0]`) |
| `deque` | fast list for adding/removing at **both ends** | queues, last-N items, sliding windows |
| `OrderedDict` | dict that remembers order + `move_to_end` | mostly old code (normal `dict` keeps order since Python 3.7) |

````python
from collections import Counter, defaultdict, namedtuple, deque, OrderedDict
````

### 1. `Counter` ⭐
| Method | What it does |
|---|---|
| `Counter(iterable)` | count each item |
| `c[item]` | count of item (**0** if missing, no KeyError) |
| `c.most_common(n)` | top n items as `(item, count)` |
| `c.update(iterable)` | add more counts |
| `c.total()` | sum of all counts (Python 3.10+) |
| `c1 + c2`, `c1 - c2` | combine / subtract counts |

````python
labels = ["cat", "dog", "cat", "bird", "cat", "dog"]
c = Counter(labels)
print(c)
print(c["cat"], c["fish"])          # missing -> 0
print(c.most_common(2))
print(c.total())

c.update(["fish", "fish"])
print(c)

print(Counter("mississippi"))       # works on strings too
print(Counter(a=3, b=1) + Counter(a=1, c=2))
````

### 2. `defaultdict` ⭐
`defaultdict(factory)` — when you access a missing key, it creates it with `factory()`:
- `defaultdict(list)` → missing key starts as `[]`
- `defaultdict(int)` → missing key starts as `0`
- `defaultdict(set)` → missing key starts as `set()`

````python
# grouping WITHOUT defaultdict
groups = {}
for word in ["apple", "avocado", "banana", "blueberry", "cherry"]:
    if word[0] not in groups:
        groups[word[0]] = []
    groups[word[0]].append(word)
print(groups)

# grouping WITH defaultdict
groups = defaultdict(list)
for word in ["apple", "avocado", "banana", "blueberry", "cherry"]:
    groups[word[0]].append(word)
print(groups)
print(dict(groups))   # convert back to a normal dict
````

````python
counts = defaultdict(int)
for ch in "hello":
    counts[ch] += 1
print(dict(counts))
````

### 3. `namedtuple`
Like a tuple (immutable, unpackable, lightweight) but with field names. Great for returning several values from a function.

````python
Point = namedtuple("Point", ["x", "y"])
p = Point(3, 4)
print(p)
print(p.x, p.y, p[0])     # by name or index
x, y = p                  # unpacking works
print(p._asdict())        # -> dict
print(p._replace(x=10))   # new tuple with one field changed (tuples are immutable)

Result = namedtuple("Result", "accuracy loss")   # fields can be a space-separated string
r = Result(0.92, 0.18)
print(f"acc={r.accuracy}, loss={r.loss}")
````

### 4. `deque` ("deck")
- `append` / `appendleft`, `pop` / `popleft` are all **O(1)** — `list.pop(0)` is slow (O(n)) on big lists.
- `deque(maxlen=n)` keeps only the **last n** items — old ones drop off automatically. ⭐

````python
d = deque([1, 2, 3])
d.append(4)
d.appendleft(0)
print(d)
print(d.pop(), d.popleft())
print(d)
d.rotate(1)       # shift right
print(d)
````

````python
last3 = deque(maxlen=3)
for i in range(6):
    last3.append(i)
    print(list(last3))
````

### 5. `OrderedDict`
Normal dicts keep insertion order already. `OrderedDict` is still handy for `move_to_end` and `popitem(last=False)`.

````python
od = OrderedDict(a=1, b=2, c=3)
od.move_to_end("a")
print(od)
print(od.popitem(last=False))   # remove the FIRST item
````

    clean = lambda s: s.replace(" ", "").lower()
    return Counter(clean(a)) == Counter(clean(b))
